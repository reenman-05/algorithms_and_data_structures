## Problem One: First Duplicate in List
**Description**  
Given a list of non-negative integers, return the first value that appears more than once. If there are multiple duplicates, return the one whose duplicate appears first in the list (see example 4). If no duplicate exists, return -1.

**Examples**
* Example 1: one duplicate
    ```python
    Input: nums = [3, 1, 4, 2, 5, 1, 6]
    Output: 1
    ```
* Example 2: no duplicate
    ```python
    Input: nums = [7, 8, 9, 10]
    Output: -1
    ```
* Example 3: multiple duplicates
    ```python
    Input: nums = [4, 5, 6, 4, 6]
    Output: 4
    ```
* Example 4: multiple duplicates  
    ```python
    Input: nums = [0, 1, 2, 1, 0]
    Output: 1
    ```

**Problem 1 Starter Code**

````python
from typing import List


def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1
test_cases = [
    [3, 1, 4, 2, 5, 1, 6],
    [7, 8, 9, 10],
    [4, 5, 6, 4, 6],
    [0, 1, 2, 1, 0],
    [5, 2, 7, 2, 5],   # my new Example A
    [0, 0],            # my new Example B
]

for nums in test_cases:
    print(f"Input: nums = {nums}")
    print(f"Output: {first_duplicate(nums)}")
    print()
````

3. Based on the provided starter code, code the solution to your assigned problem here in Python. Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Try to find the solution with best time and space complexity.

````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    myset = set()
    for num in nums:
        if num in myset:
            return num
        myset.add(num)
    return -1

# Tests
assert first_duplicate([3, 1, 4, 2, 5, 1, 6]) == 1
assert first_duplicate([7, 8, 9, 10]) == -1
assert first_duplicate([4, 5, 6, 4, 6]) == 4
assert first_duplicate([0, 1, 2, 1, 0]) == 1
assert first_duplicate([5, 2, 7, 2, 5]) == 2
assert first_duplicate([0, 0]) == 0
````


4. Walk through your Python code in plain words and explain why your solution works.


#In the above code, I create an empty set called myset to remember what I have gone through. 
Then I go through the list one by one at a time, from left to right.
Now in the For loop, each number I check whether it’s already in the myset. If it is, it will be second time I’ve encountered it, so I return it immediately. 
If it isn’t, I add it to myset and move on.
If I reach the end of the list without returning meaning no number repeated, I return -1.

#why your solution works.
Here the answer is the number whose second appearance comes earliest. I read the list from left to right and keep a set of the numbers I’ve already seen. The first time I find a number that’s already in the set, I return it. Since I go in order, no repeat can show up earlier than that one. A set is a good fit because checking “have I seen this before?” is very fast.
