## Problem Two: Path to Leaves
**Description**  
Given the `root` of a binary tree, return all root to leaf paths in any order.

**Examples**

* Example 1:

    ![](./images/q1_ex1.png)

    Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?* It is a level-order (breadth-first)serialization. Nodes are listed level by level, left to right

    Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

* Example 2:

    ![](./images/q1_ex3.png)

    Input: `root = [10, 9, 7, 8]`

    Output: [[10, 7], [10, 9, 8]]

**Problem 2 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
from typing import List
def bt_path(root: TreeNode) -> List[List[int]]:
  result = []
  path = []

  def dfs(node):
        if not node:
            return
        path.append(node.val)

        if not node.left and not node.right:
            result.append(path[:])
        else:
            dfs(node.left)
            dfs(node.right)

        path.pop()

  dfs(root)
  return result
```

Example: Input:[-2, 0, 4, 0, 1] -> Output: [-2, 4, 1, 0, 0]
We are going left to right, each non-zero number is copied to the next free spot at the front, and zeros are skipped:

-2 goes to position 0

The remaining positions (3 and 4) are filled with zeros, giving [-2, 4, 1, 0, 0].


4. Copy the solution your partner wrote. 


```python
# Your answer here
from typing import List
def move_zeros_to_end(nums: List[int]) -> List[int]:
    write = 0

    # Move each nonzero number to the next available position.
    for value in nums:
        if value != 0:
            nums[write] = value
            write += 1

    # Set all remaining positions to zero.
    while write < len(nums):
        nums[write] = 0
        write += 1

    return nums
```
